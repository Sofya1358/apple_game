### Шаг 1. Переменные

```python
import tkinter as tk

window = tk.Tk()
window.title("Переменные")
window.geometry("400x300")

name = "Вика"
age = 10

tk.Label(window, text=name).pack()
tk.Label(window, text=age).pack()

window.mainloop()
```

```python
import tkinter as tk

window = tk.Tk()
window.title("Типы данных")
window.geometry("400x300")

name = "Вика"       # str
age = 14            # int
height = 1.65       # float
is_student = True   # bool

tk.Label(window, text=f"{name} ({type(name).__name__})").pack()
tk.Label(window, text=f"{age} ({type(age).__name__})").pack()
tk.Label(window, text=f"{height} ({type(height).__name__})").pack()
tk.Label(window, text=f"{is_student} ({type(is_student).__name__})").pack()

window.mainloop()
```

```python
import tkinter as tk
# Создаем окно
window = tk.Tk()
window.title("Пользовательский ввод")
window.geometry("400x300")

# Создаем поле ввода имени
tk.Label(window, text="Твоё имя:").pack()
entry_name = tk.Entry(window)
entry_name.pack()

# Создаем поле ввода возраста
tk.Label(window, text="Твой возраст:").pack()
entry_age = tk.Entry(window)
entry_age.pack()

button = tk.Button(
    window,
    text="Узнать"
)
button.pack()

window.mainloop()
```

### Шаг 4. Цикл

```python
import tkinter as tk


# -------------------------
# Проверяем возраст
# -------------------------

def check_age():
    name = entry_name.get()
    age = int(entry_age.get())

    if age < 12:
        result = f" Тебе {age}.\n До {age}-летия ты дошёл,\n а до официального взросления осталось ещё целых {18 - age} лет.\n Наслаждайся детством!"

    elif age == 12:
        result = " Тебе 12 лет.\n Ровно половина пути до 24-х!\n А до совершеннолетия осталось ждать ещё 6 лет."

    elif age == 13:
        result = " Тебе 13 лет.\n До совершеннолетия осталось 5 лет\n  — \n это ровно столько,\n  сколько длится стандартное обучение в университете!"

    elif age == 14:
        result = " Тебе 14 лет.\n До совершеннолетия осталось 4 года.\n Кстати, это ровно один олимпийский цикл!"

    elif age == 15:
        result = " Тебе 15 лет.\n Осталось всего 3 года до совершеннолетия!\n  Время полетит очень быстро, успей определиться с хобби."

    elif age == 16:
        result = " Тебе 16 лет.\n До 18-летия осталось всего 2 года.\n Скоро можно будет сдавать на права!"

    elif age == 17:
        result = " Тебе 17 лет.\n Финишная прямая! Всего 1 год до совершеннолетия.\n Последний год, когда можно официально побыть ребёнком."

    elif age == 18:
        result = " Поздравляем!\n Тебе 18 лет — совершеннолетие наступило прямо сейчас!\n Добро пожаловать во взрослую жизнь."

    else:
        result = f" Тебе {age}.\n Ты совершеннолетний уже {age - 18} лет!"

    result_label.config(text=result)

    # Показываем дополнительные кнопки
    # только если человеку меньше 18 лет
    if age < 18:
        future_button.pack(pady=5)
        childhood_button.pack(pady=5)
    else:
        future_button.pack_forget()
        childhood_button.pack_forget()


# -------------------------
# Прогноз на 5 лет
# -------------------------

def future():
    name = entry_name.get()
    age = int(entry_age.get())

    result = f"Прогноз для {name}:\n\n"

    for years in range(1, 6):

        future_age = age + years

        if years == 1:
            years_word = "год"
        elif 2 <= years <= 4:
            years_word = "года"
        else:
            years_word = "лет"

        if future_age < 18:
            status = "ты ещё несовершеннолетний"
        else:
            status = "ты уже совершеннолетний"

        result += (
            f"Через {years} {years_word} "
            f"тебе будет {future_age} лет — {status}.\n"
        )

    result_label.config(text=result)


# -------------------------
# Прогноз до совершеннолетия
# -------------------------

def childhood():
    name = entry_name.get()
    age = int(entry_age.get())

    result = f"Как будет меняться возраст {name}:\n\n"

    future_age = age + 1

    while future_age <= 18:

        years = future_age - age

        if years == 1:
            years_word = "год"
        elif 2 <= years <= 4:
            years_word = "года"
        else:
            years_word = "лет"

        if future_age < 18:
            status = "ты ещё несовершеннолетний"
        else:
            status = "ты станешь совершеннолетним"

        result += (
            f"Через {years} {years_word} "
            f"тебе будет {future_age} лет — {status}.\n"
        )

        future_age += 1

    result_label.config(text=result)


# -------------------------
# Создаём окно
# -------------------------

window = tk.Tk()
window.title("Кто ты по возрасту?")
window.geometry("550x450")


# Имя
tk.Label(window, text="Твоё имя:").pack()

entry_name = tk.Entry(window)
entry_name.pack()


# Возраст
tk.Label(window, text="Твой возраст:").pack()

entry_age = tk.Entry(window)
entry_age.pack()

# Создаем пустое место на экране, куда функция check_age запишет ответ
result_label = tk.Label(window, text="")
result_label.pack()

# Основная кнопка
button = tk.Button(
    window,
    text="Узнать",
    command=check_age
)
button.pack(pady=10)

# Дополнительные кнопки
# Пока они скрыты

future_button = tk.Button(
    window,
    text="Прогноз на 5 лет",
    command=future
)

childhood_button = tk.Button(
    window,
    text="Прогноз до совершеннолетия",
    command=childhood
)


window.mainloop()
```
